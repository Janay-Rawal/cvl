```python


from keras.applications import DenseNet121
model = DenseNet121(weights='imagenet', include_top=False, pooling='avg', input_shape=(32, 32, 3))


from keras.datasets import cifar10
from keras.applications.densenet import preprocess_input
import numpy as np
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt


(X_train, y_train), (X_test, y_test) = cifar10.load_data()

# Smaller subset for faster training
X_train, y_train = X_train[:200], y_train[:200]
X_test, y_test = X_test[:50], y_test[:50]

# Preprocess to match ImageNet input scaling
X_train = preprocess_input(X_train.astype('float32'))
X_test = preprocess_input(X_test.astype('float32'))

features_train = model.predict(X_train, verbose=1)
features_test = model.predict(X_test, verbose=1)


clf = SVC(kernel='linear')
clf.fit(features_train, y_train.ravel())


y_pred = clf.predict(features_test)
acc = accuracy_score(y_test, y_pred)
print("Accuracy:", acc)

# Confusion matrix
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap='Blues')
plt.title("Confusion Matrix")
plt.show()


accuracies = []
for i in range(20, 201, 20):
    clf = SVC(kernel='linear')
    clf.fit(features_train[:i], y_train[:i].ravel())
    y_pred = clf.predict(features_test)
    accuracies.append(accuracy_score(y_test, y_pred))

plt.plot(range(20, 201, 20), accuracies, marker='o')
plt.xlabel("Training samples (simulated epochs)")
plt.ylabel("Accuracy")
plt.title("Accuracy vs Training Size (simulated Epochs)")
plt.grid(True)
plt.show()
```
